```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from keras.models import Sequential
from keras.layers import Dense
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv("crop (1).csv")
print(data)
```

```python
sns.heatmap(data.isnull())
```

```python
scrop=data[(data['temperature']>35)]['label'].unique()
print(scrop)
```

```python
wcrop=data[(data['temperature']<20)]['label'].unique()
print(wcrop)
```

```python
mcrop=data[(data['rainfall']>200)]['label'].unique()
print(mcrop)
```

```python
x=data.iloc[:,:-1].values
y=data.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
y=le.fit_transform(y)
y_train=le.fit_transform(y_train)
y_test=le.transform(y_test)
```

```python
print(le.classes_)
```

```python
model=Sequential()
model.add(Dense(128,activation='relu',input_shape=(x_train.shape[1],)))
model.add(Dense(64,activation='relu'))
model.add(Dense(32,activation='relu'))
model.add(Dense(16,activation='relu'))
model.add(Dense(len(le.classes_),activation='softmax'))
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history=  model.fit(x_train,y_train,epochs=50,batch_size=16,validation_split=.2)
```

```python
from os import access
loss,accuracy=model.evaluate(x_test,y_test)
print("Loss:",loss)
print("Accuracy:",accuracy)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('model accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['Train','Validation'],loc='upper left')
plt.show()
```

```python
import joblib

# Save the Keras deep learning model
model.save('crop_model.keras')

# Save the label encoder (maps output numbers to crop names like 'rice', 'watermelon')
joblib.dump(le, 'label_encoder.pkl')
```
